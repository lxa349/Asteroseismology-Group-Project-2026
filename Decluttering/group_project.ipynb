```python
import astropy
import numpy as np
from pathlib import Path
from astropy.nddata import CCDData
from astropy.io import fits
import astropy.units as u
import matplotlib.pyplot as plt
import warnings
import glob
```

```python
#PIC = fits.util.get_testdata_filepath('data/LOPS2PICtarget2.1.0.1-t-fg-c-scv.fits')
PIC = fits.open('data/LOPS2PICtarget2.1.0.1-t-fg-c-scv.fits')
#headers = np.asarray(PIC[1].header)
data = np.asarray(PIC[1].data)
print(len(data["PICid"]))
```

```python
myepictable = []

for i in range(len(data["Mass"])):
    if data["Mass"][i] < 1.5:
        myepictable.append(data[i])

data_reduced = np.asarray(myepictable)
```

```python
print(f"{data_reduced["Mass"][0]} solar masses, {data_reduced["Radius"][0]} solar radii, {data_reduced["Teff"][0]} K")
```

```python
def characteristic_timescale_granulation(i):
    tau_gran = 250
    M = data_reduced["Mass"][i]
    R = data_reduced["Radius"][i]
    T = data_reduced["Teff"][i]
    dM = M / (1.989 * 10**30) #kg
    dR = R / 696340 #km
    dT = T / 5778 #kelvin

    return( (M * R**(-2) * dT**(-0.5))**(-1) * tau_gran)
```

```python
print(f"{characteristic_timescale_granulation(0)} seconds")
```
