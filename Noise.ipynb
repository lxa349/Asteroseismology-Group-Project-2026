```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits


def random(values):
    randomised = []
    for i in values:
        a=np.random.uniform(0,1)
        randomised.append(i*-np.log(a))
    return randomised

def add_whitenoise(data, noise_lvl):
    noise_lvl=[noise_lvl]*len(data)
    return np.array(data)+np.array(random(noise_lvl))
```

```python
x = np.linspace(0,10,10000)
y = random(np.sin(x))

y_n = add_whitenoise(y,1)

plt.plot(x,y,alpha=0.5)
plt.plot(x,y_n,alpha=0.5)
```
